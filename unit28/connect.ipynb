## 6. GPU

```python
import importlib.util
import os
import subprocess

colab = os.getcwd() == "/content" and importlib.util.find_spec("google.colab") is not None
try:
    gpu = subprocess.run(["nvidia-smi"], capture_output=True).returncode == 0
except FileNotFoundError:
    gpu = False
print("○" if colab else "×", "Colab のサーバーで動いている")
print("○" if gpu else "×", "GPU が使える")
```
